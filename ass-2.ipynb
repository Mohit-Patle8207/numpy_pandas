````python
# **Q1.** Create the array `temperatures = [36.5, 37.2, 38.9, 36.1, 39.5, 37.8, 40.1, 36.8]`.  
# - Find all temperatures **above 38.0** (fever threshold).  
# - Count how many readings are **normal** (between 36.0 and 37.5).  
# - What percentage of readings are **high fever** (above 39.0)?

temperatures = np.array([36.5, 37.2, 38.9, 36.1, 39.5, 37.8, 40.1, 36.8])
above_38 = temperatures[temperatures > 38.0]
normal_count=np.sum((temperatures >= 36.0) & (temperatures <= 37.5))
high_fever_percentage=np.sum(temperatures > 39.0)/temperatures.size*100

print("temperatures above 38.0 ", above_38)
print("count of normal readings ", normal_count)    
print("percentage of high fever readings ", high_fever_percentage)
````

````python
# **Q1.** Create an array `[1.0, np.nan, 3.0, np.nan, 5.0, np.nan, 7.0]`.  
# - Count how many values are NaN.  
# - Compute the mean using both `np.mean()` and `np.nanmean()`. Compare the results.  
# - Replace all NaN values with the **median** of the non-NaN values.

arr = np.array([1.0, np.nan, 3.0, np.nan, 5.0, np.nan, 7.0])
nan_count = np.isnan(arr).sum()
print("Number of NaN values: ", nan_count)

mean_with_nan = np.mean(arr)
mean_without_nan = np.nanmean(arr)
print("Mean with NaN: ", mean_with_nan)
print("Mean without NaN: ", mean_without_nan)

median_value = np.nanmedian(arr)
arr_filled = np.where(np.isnan(arr), median_value, arr)
print("Array after filling NaN with median: ", arr_filled)
````

````python
# **Q2.** Given `scores = [55, 70, 82, 91, 64, 78, 88, 45, 73, 60]`:
# - Compute `mean`, `std`, and `median`.
# - Manually compute **Z-scores**: `(scores - mean) / std`.
# - Which students have Z-score **above 1.0** (high performers)?


scores = np.array([55, 70, 82, 91, 64, 78, 88, 45, 73, 60])
mean_score = np.mean(scores)
std_score = np.std(scores)
median_score = np.median(scores)
print("Mean: ", mean_score)
print("Standard Deviation: ", std_score)
print("Median: ", median_score)
z_scores = (scores - mean_score) / std_score
print("Z-scores: ", z_scores)
high_performers = np.where(z_scores > 1.0)[0]+1
print("Students with Z-score above 1.0 (high performers): ", high_performers)
````
