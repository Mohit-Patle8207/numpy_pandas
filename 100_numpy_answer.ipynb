````python
# #### 63. Create an array class that has a name attribute (★★☆)
# `hint: class method`

class Array(np.ndarray):

    def __new__(cls, input_array, name=""):
        obj = np.asarray(input_array).view(cls)
        obj.name = name
        return obj

    def __array_finalize__(self, obj):
        if obj is None:
            return
        self.name = getattr(obj, 'name', "")

x=Array([1,2,3],name="my array")
print(x)
print(x.name)
y=Array([2,3,6])
print(y)
print(y.name)
````

````python
# #### 80. Consider an arbitrary array, write a function that extracts a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)
# `hint: minimum maximum`

def extract_centered(A, center, shape, fill=0):
    """
    Extract a centered sub-array with padding.

    Parameters:
    A      : input array (2D)
    center : (i, j) center position
    shape  : (h, w) output shape
    fill   : padding value
    """
    H, W = A.shape
    h, w = shape
    ci, cj = center

    # output array
    out = np.full((h, w), fill, dtype=A.dtype)

    # compute half-window
    hi = h // 2
    hj = w // 2

    # bounds in A
    i_min = max(ci - hi, 0)
    i_max = min(ci + hi + 1, H)

    j_min = max(cj - hj, 0)
    j_max = min(cj + hj + 1, W)

    # corresponding region in output
    out_i_min = hi - (ci - i_min)
    out_i_max = out_i_min + (i_max - i_min)

    out_j_min = hj - (cj - j_min)
    out_j_max = out_j_min + (j_max - j_min)

    # copy
    out[out_i_min:out_i_max, out_j_min:out_j_max] = A[i_min:i_max, j_min:j_max]

    return out
````

````python

# #### 85. Create a 2D array subclass such that Z[i,j] == Z[j,i] (★★★)
# `hint: class method`

class SymmetricMatrix(np.ndarray):
    
    def __new__(cls, input_array):
        obj = np.asarray(input_array).view(cls)
        return obj

    def __setitem__(self, index, value):
        i, j = index

        super().__setitem__((i, j), value)
        super().__setitem__((j, i), value)
````

````python
# #### 88. How to implement the Game of Life using numpy arrays? (★★★)
# `No hints provided...`

import numpy as np

def game_of_life(board):
    # Count neighbors using shifted views
    neighbors = (
        np.roll(np.roll(board,  1, 0),  1, 1) +
        np.roll(np.roll(board,  1, 0),  0, 1) +
        np.roll(np.roll(board,  1, 0), -1, 1) +
        np.roll(np.roll(board,  0, 0),  1, 1) +
        np.roll(np.roll(board,  0, 0), -1, 1) +
        np.roll(np.roll(board, -1, 0),  1, 1) +
        np.roll(np.roll(board, -1, 0),  0, 1) +
        np.roll(np.roll(board, -1, 0), -1, 1)
    )

    return (
        ((board == 1) & ((neighbors == 2) | (neighbors == 3))) |
        ((board == 0) & (neighbors == 3))
    ).astype(int)

board = np.array([
    [0, 1, 0],
    [0, 1, 0],
    [0, 1, 0]
])

print("Current:")
print(board)

next_board = game_of_life(board)
````

````python
# #### 90. Given an arbitrary number of vectors, build the cartesian product (every combination of every item) (★★★)
# `hint: np.indices`
def cartesian_product(*arrays):
    shapes = [len(a) for a in arrays]
    
    grids = np.indices(shapes)
    
    # reshape each grid into a column
    product = np.stack([
        arrays[i][grids[i].ravel()] 
        for i in range(len(arrays))
    ], axis=1)
    
    return product
A = np.array([1, 2])
B = np.array([3, 4])
C = np.array([5, 6])

result = cartesian_product(A, B, C)

print(result)
````

````python
# #### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
# `hint: np.cumsum, np.interp`
X = np.array([0, 1, 2, 4, 7])
Y = np.array([0, 2, 1, 3, 2])

# Distance between consecutive points
d = np.sqrt(np.diff(X)**2 + np.diff(Y)**2)

# Cumulative distance along path
s = np.concatenate(([0], np.cumsum(d)))

# Number of desired equidistant samples
n_samples = 20

# Equally spaced distances
s_uniform = np.linspace(0, s[-1], n_samples)

# Interpolate coordinates
X_uniform = np.interp(s_uniform, s, X)
Y_uniform = np.interp(s_uniform, s, Y)

print(X_uniform)
print(Y_uniform)
````

````python
# #### 100. Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means). (★★★)
# `hint: np.percentile`

X = np.random.randn(100)  # example data

n_boot = 10000  # number of bootstrap samples

# 1. bootstrap resampling (each row is a resample)
boot_samples = np.random.choice(X, size=(n_boot, len(X)), replace=True)

# 2. compute mean of each bootstrap sample
boot_means = np.mean(boot_samples, axis=1)

# 3. 95% confidence interval
ci_lower, ci_upper = np.percentile(boot_means, [2.5, 97.5])

print("Bootstrap 95% CI:", (ci_lower, ci_upper))
````
